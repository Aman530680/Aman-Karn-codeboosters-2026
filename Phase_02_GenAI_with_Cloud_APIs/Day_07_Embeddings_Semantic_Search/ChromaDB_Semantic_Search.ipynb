```python
smart_notes = chroma_client.get_or_create_collection(
    name="college_notes"
)

print("Collection Created Successfully!")
print("Collection Name: college_notes")
```

```python
notes = [
    "Neural networks learn patterns from data",
    "Supervised learning uses labeled examples",
    "Unsupervised learning finds hidden patterns",
    "Deep learning powers modern AI systems",
    "Machine learning improves prediction accuracy",

    "Python supports object oriented programming",
    "Functions help organize code",
    "Loops automate repetitive tasks",
    "Lists store multiple values",
    "Dictionaries store key value pairs",

    "SQL is used to query databases",
    "Primary keys uniquely identify records",
    "Normalization reduces redundancy",
    "Indexes improve query performance",
    "Database transactions ensure consistency"
]

print("Total Notes:", len(notes))
```

```python
ids = [
    "doc1","doc2","doc3","doc4","doc5",
    "doc6","doc7","doc8","doc9","doc10",
    "doc11","doc12","doc13","doc14","doc15"
]

print("Total IDs:", len(ids))
```

```python
metadatas = [
    {"subject":"Machine Learning"},
    {"subject":"Machine Learning"},
    {"subject":"Machine Learning"},
    {"subject":"Machine Learning"},
    {"subject":"Machine Learning"},

    {"subject":"Programming"},
    {"subject":"Programming"},
    {"subject":"Programming"},
    {"subject":"Programming"},
    {"subject":"Programming"},

    {"subject":"Database"},
    {"subject":"Database"},
    {"subject":"Database"},
    {"subject":"Database"},
    {"subject":"Database"}
]

print("Metadata Created!")
```

```python
smart_notes.add(
    documents=notes,
    ids=ids,
    metadatas=metadatas
)

print("15 Notes Added Successfully!")
print("Total Notes:", smart_notes.count())
```

```python
query = "How do computers learn from examples?"

results = smart_notes.query(
    query_texts=[query],
    n_results=3
)

print("QUERY:", query)
print("=" * 60)

for doc in results["documents"][0]:
    print(doc)
```

```python
print("TOP RESULTS")
print("=" * 60)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f}")
    print(doc)
    print("-" * 40)
```

```python
query = "computer learning"

print("KEYWORD SEARCH")
print("=" * 40)

for note in notes:
    if query.lower() in note.lower():
        print(note)
```

```python
results = smart_notes.query(
    query_texts=[query],
    n_results=3
)

print("SEMANTIC SEARCH")
print("=" * 40)

for doc in results["documents"][0]:
    print(doc)
```

```python
print("KEYWORD SEARCH vs SEMANTIC SEARCH")
print("=" * 60)

query = "computer learning"

print("\nKEYWORD SEARCH:")
found = False

for note in notes:
    if query.lower() in note.lower():
        print(note)
        found = True

if not found:
    print("No exact matches found")

print("\nSEMANTIC SEARCH:")

results = smart_notes.query(
    query_texts=[query],
    n_results=3
)

for doc in results["documents"][0]:
    print(doc)
```
