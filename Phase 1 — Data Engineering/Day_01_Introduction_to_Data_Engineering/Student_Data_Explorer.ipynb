```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
student_data = pd.read_csv("student_performance.csv")
student_data["total_score"] = student_data["math_score"] + student_data["programming_score"]
print("STUDENT DATA EXPLORER")
print()


print("Dataset Overview")
total_students = len(student_data)
print("Total Students :", total_students)
total_columns = len(student_data.columns)
print("Total Columns :", total_columns)
unique_departments = student_data["department"].nunique()
print("Unique Departments :", unique_departments)
print()


print("Department Wise Analysis")
department_count = student_data["department"].value_counts()
print(department_count)
print()


print("Score Analysis")
highest_math = student_data["math_score"].max()
print("Highest Math Score :", highest_math)
lowest_math = student_data["math_score"].min()
print("Lowest Math Score :", lowest_math)
average_math = student_data["math_score"].mean()
print("Average Math Score :", average_math)
average_programming = student_data["programming_score"].mean()
print("Average Programming Score :", average_programming)
average_attendance = student_data["attendance_percentage"].mean()
print("Average Attendance Percentage :", average_attendance)
print()



print("Gender Wise Analysis")
gender_total = student_data.groupby("gender")["total_score"].mean()
print(gender_total)
print()



print("Top 5 Students")
top_students = student_data.sort_values(by="total_score", ascending=False)
print(top_students[["name", "department", "total_score"]].head(5))
print()



print("Attendance Analysis")
low_attendance = student_data[student_data["attendance_percentage"] < 75]
if len(low_attendance) > 0:
    print(low_attendance[["name", "department", "attendance_percentage"]])
else:
    print("All students have attendance above 75%")
```
