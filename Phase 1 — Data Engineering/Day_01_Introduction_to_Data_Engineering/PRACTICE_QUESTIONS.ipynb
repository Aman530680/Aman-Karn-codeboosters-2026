```python
#How many students are from the Electronics Department?

import pandas as pd
student_data = pd.read_csv("student_performance.csv")
electronics_students = student_data[student_data["department"] == "Electronics"]
print("Number of students in Electronics Department :")
print(len(electronics_students))
```

```python
#What is the average attendance of female students?

import pandas as pd
student_data = pd.read_csv("student_performance.csv")
female_attendance = student_data[student_data["gender"] == "Female"]["attendance_percentage"].mean()
print("Average attendance of female students :")
print(female_attendance)
```

```python
#Who is the student with the lowest programming score?

import pandas as pd
student_data=pd.read_csv("student_performance.csv")
lowest_programming=student_data[student_data["programming_score"]==student_data["programming_score"].min()]
print("Student with lowest programming score:")
print(lowest_programming[["name","programming_score"]])
```

```python
#How many students scored above 80 in math?

import pandas as pd
student_data = pd.read_csv("student_performance.csv")
math_above_80 = student_data[student_data["math_score"]>80]
print("Number of students scored above 80 in math:")
print(len(math_above_80))
```

```python
#What is the average total score for students with attendance above 90%?

import pandas as pd
student_data = pd.read_csv("student_performance.csv")
student_data["total_score"] = student_data["math_score"] + student_data["programming_score"]
attendance_above_90 = student_data[student_data["attendance_percentage"]>90]["total_score"].mean()
print("Average total score for students with attendance above 90%:")
print(attendance_above_90)
```
