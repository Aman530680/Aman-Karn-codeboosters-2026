```python
df['average_score'] = df[['math_score', 'science_score', 'english_score', 'programming_score']].mean(axis=1)
lowest_performing_student = df.loc[df['average_score'].idxmin()]
print("Lowest Performing Student:")
print(f"Name: {lowest_performing_student['name']}")
print(f"Age: {lowest_performing_student['age']}")
print(f"Gender: {lowest_performing_student['gender']}")
print(f"Department: {lowest_performing_student['department']}")
print(f"Average Score: {lowest_performing_student['average_score']:.2f}")
```
