```python
from pyspark.sql import SparkSession

print("Spark Imported Successfully!")
```

```python
spark = SparkSession.builder \
    .appName("Sales Analysis") \
    .getOrCreate()

print("Spark Session Created!")
```

```python
df = spark.read.csv(
    "large_sales_data.csv",
    header=True,
    inferSchema=True
)

print("Dataset Loaded Successfully!")
```

```python
df.show(5)
```

```python
print(df.columns)
```

```python
df.printSchema()
```

```python
print("Total Rows:", df.count())
```

```python
df.select(
    "customer_name",
    "product",
    "revenue"
).show(10)
```

```python
df.filter(
    df.revenue > 100000
).show()
```

```python
df.groupBy(
    "region"
).count().show()
```

```python
from pyspark.sql.functions import sum

df.groupBy(
    "region"
).agg(
    sum("revenue").alias("total_revenue")
).show()
```

```python
df.groupBy(
    "product"
).count().orderBy(
    "count",
    ascending=False
).show()
```

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
    .appName("Medallion Architecture") \
    .getOrCreate()

print("Spark Session Created!")
```

```python
df_bronze = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("large_sales_data.csv")

df_bronze.show(5)
```

```python
print("Bronze Layer Row Count:", df_bronze.count())
```

```python
df_silver = df_bronze.dropDuplicates()
```

```python
df_silver = df_silver.fillna("Unknown")
```

```python
df_silver.show(5)

print("Silver Layer Row Count:", df_silver.count())
```

```python
from pyspark.sql.functions import sum

df_gold = df_silver.groupBy("region") \
    .agg(
        sum("revenue").alias("total_revenue")
    )

df_gold.show()
```

```python
print("=" * 60)

print("MEDALLION ARCHITECTURE COMPLETED")

print("=" * 60)

print("Bronze Layer : Raw Data")

print("Silver Layer : Cleaned Data")

print("Gold Layer : Business Ready Data")
```
