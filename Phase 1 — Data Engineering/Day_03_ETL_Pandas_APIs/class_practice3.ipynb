```python
import pandas as pd
df = pd.read_csv('messy_sales_data.csv')
initial_df_rows = len(df)
print(f"Dataset loaded : {df.shape[0]} student, {df.shape[1]} column, ")
print (f"columns : {df.columns.to_list()}")
print(f"first five row ")
df.head(5)
```

```python
print(f"Dataset consists of {df.shape[0]} rows and {df.shape[1]} columns")
print(f"Columns : {df.columns.to_list()}")
print("First 5 rows:")
print(df.head())
```

```python
print("=" * 55)
print(" DATA QUALITY DIAGONOSIS REPORT")
print("=" * 55)

# 1 missing values
print("\nMissing values per column:")
print(df.isnull().sum())

# 2 Duplicates
print("\nDuplicate rows :", df.duplicated().sum())

# 3 DataTypes
print("\nDataTypes :")
print(df.dtypes)

# 4 Unique values in text columns
print("\nUnique Categories:", df['category'].unique())
print("\nSample customer name: ", df['customer_name'].dropna().unique()[:8])
print("\nsample order values:", df['order_date'].unique()[:6])
```

```python
# Before drop
print(f"Before dropping duplicates: {len(df)} rows")

# Count duplicates
duplicate_rows_count = df.duplicated().sum()
print(f"Number of duplicate rows found: {duplicate_rows_count}")

# Show duplicates
if duplicate_rows_count > 0:
    print("\nIdentified duplicate rows (keeping all occurrences for display):")
    # All duplicate occurrences
    print(df[df.duplicated(keep=False)])

# Define subset
columns_for_uniqueness = ['order_id', 'customer_name', 'product', 'order_date']

# Drop duplicates
df.drop_duplicates(subset=columns_for_uniqueness, inplace=True)

# After drop
print(f"\nAfter dropping duplicates: {len(df)} rows")
```

```python
# Date processing

# Original dates
print('Sample data before parsing:')
print(df['order_date'].head(8).tolist())

# Convert dates
df["order_date"] = pd.to_datetime(
    df["order_date"],
    errors="coerce"
)

# Count NaT
nat_count = df['order_date'].isnull().sum()
print(f'\nUnparseable data (NaN count): {nat_count}')

# Extract parts
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()

# Parsed data
print('\nSample data after parsing:')
print(df[['order_date', 'year', 'month', 'month_name']].head(5))
```

```python
print('Before standaridazation:', df['customer_name'].unique()[:8])

df['customer_name'] = (
    df['customer_name']
    .str.strip() # remove space ' Ram ' to 'Ram'
    .str.title() # Convert to title case: 'bibek dahal' to 'Bibek Dahal'
    )

print('\nAfter standardization:', df['customer_name'].unique()[:8])

print(f'\n Before Keyboard rows with Electronics category:')
wrong_mask = (df['product'] == 'Keyboard') & (df['category'] == 'Electronics')
print(df[wrong_mask][['product','category']])

df.loc[wrong_mask, 'category'] = 'Accessories'

print(f'\nAfter fix: Unique categories:', df['category'].unique())
```

```python
# Convert 'quantity' to numeric, coercing errors to NaN
df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')

# Convert 'unit_price' to numeric, coercing errors to NaN
df['unit_price'] = pd.to_numeric(df['unit_price'], errors='coerce')

# Calculate 'revenue' by multiplying 'quantity' and 'unit_price'
df['revenue'] = df['quantity'] * df['unit_price']

print("Quantity, Unit Price, and Revenue columns processed.")
print(df[['quantity', 'unit_price', 'revenue']].head())
print(df[['customer_name', 'product', 'quantity', 'unit_price', 'revenue']].head())
print(f"\nTotal Revenue across all orders: {df['revenue'].sum():.0f}")
```

```python
print('=' * 55)
print(' POST-CLEANING VALIDATION REPORT')
print('=' * 55)

# Using initial_df_rows stored from the data loading cell
if 'initial_df_rows' not in locals():
    print("Warning: 'initial_df_rows' not found. Assuming initial row count was 30 for demonstration.")
    original_row_count = 30
else:
    original_row_count = initial_df_rows

print(f'Original rows: {original_row_count}')
print(f'Cleaned rows: {len(df)}')
print(f'Rows removed: {original_row_count - len(df)} (likely duplicates/NaNs)')

print(f'Total missing values: {df.isnull().sum().sum()}')
print(f'Duplicates: {df.duplicated().sum()}')
print(f'Date nulls: {df["order_date"].isnull().sum()}')
print(f'Revenue NaN: {df["revenue"].isnull().sum()}')
print(f'Categories: {sorted(df["category"].unique().astype(str))}')
print('=' * 55)

all_clean = (
    df.isnull().sum().sum() == 0 and
    df.duplicated().sum() == 0
)
print (f"Data is clean: {all_clean}")
```

```python
product_revenue = (
    df.groupby('product')['revenue']
    .sum()
    .sort_values(ascending=False)
)
print("Revenue per product (sorted descending):")
print(product_revenue)
```

```python
product_rev = (
    df.groupby('product')['revenue']
    .sum()
    .reset_index()
    .sort_values('revenue', ascending=False)
)

print('Total revenue by product:')
print(product_rev.to_string(index=False))

category_summary = df.groupby('category').agg(
    total_revenue = ('revenue', 'sum'),
    avg_order_value = ('revenue', 'mean'),
    num_orders = ('order_id', 'count'),
    unique_products = ('product', 'nunique')
).round(2).reset_index()

print('\nCategory Summary:')
print(category_summary.to_string(index=False))
```

```python
df.to_csv('clean_sales_data.csv', index=False)

print('Cleaned data saved to: clean_sales_data.csv')
print(f'Final dataset: {df.shape[0]} rows x {df.shape[1]} columns')
print('\nETL Pipeline for Sales Data: COMPLETE')
print('EXTRACT -> messy_sales_data.csv loaded')
print('TRANSFORM -> cleaned_sales_data.csv saved')
print('LOAD -> clean_sales_data.csv saved')
```
