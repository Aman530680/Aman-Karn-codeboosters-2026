```python
#Q4: What is the difference between fillna(e) and fillna(df['col'].median()) ? When would you prefer each?
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'numerical_col': [10, 20, np.nan, 40, 50, np.nan, 70],
    'categorical_col': ['A', 'B', 'C', np.nan, 'E', 'F', np.nan]
})

print("Original DataFrame:\n", df)

# fillna(e) - filling with a static value
df_filled_static = df.copy()
df_filled_static['numerical_col'] = df_filled_static['numerical_col'].fillna(0)
df_filled_static['categorical_col'] = df_filled_static['categorical_col'].fillna('UNKNOWN')
print("\nDataFrame after fillna(e) with static values (0 for numerical, 'UNKNOWN' for categorical):\n", df_filled_static)
print("Use fillna(e) when you have a specific, predetermined value to replace missing data, like a default or a sentinel value.")

# fillna(df['col'].median()) - filling with a statistical measure
df_filled_median = df.copy()
median_val = df_filled_median['numerical_col'].median()
df_filled_median['numerical_col'] = df_filled_median['numerical_col'].fillna(median_val)
print(f"\nDataFrame after fillna(df['col'].median()) for numerical_col (median={median_val}):\n", df_filled_median)
print("Use fillna(df['col'].median()) (or mean/mode) when imputing missing numerical data with a central tendency measure from the data itself. Median is preferred over mean when data might have outliers.")
```
